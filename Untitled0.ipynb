```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from IPython.display import Audio, display
import warnings
warnings.filterwarnings('ignore')

print(f"PyTorch versión: {torch.__version__}")
print(f"Entorno: {'GPU' if torch.cuda.is_available() else 'CPU'}")
print()

# ============================================================================
# FASE A: TRANSDUCCIÓN
# ============================================================================
print("="*80)
print("FASE A: TRANSDUCCIÓN")
print("="*80)

# Generación de imagen sintética RGB con gradientes
img_tensor = torch.zeros(3, 256, 256)
x = torch.linspace(0, 1, 256)
y = torch.linspace(0, 1, 256)
xx, yy = torch.meshgrid(x, y, indexing='ij')

img_tensor[0] = xx
img_tensor[1] = yy
img_tensor[2] = 0.5 * (xx + yy)

print(f"✅ Imagen Tensor: {img_tensor.shape}")
print(f"Dimensiones: [C={img_tensor.shape[0]}, H={img_tensor.shape[1]}, W={img_tensor.shape[2]}]")
print()

sample_rate = 16000
duration = 2.0
frequency = 440.0
t = np.linspace(0, duration, int(sample_rate * duration), endpoint=False)
audio_signal = 0.5 * np.sin(2 * np.pi * frequency * t)
audio_stereo = np.vstack([audio_signal, 0.5 * np.sin(2 * np.pi * frequency * t + np.pi/4)])
audio_tensor = torch.from_numpy(audio_stereo).float()

print(f"✅ Audio Tensor: {audio_tensor.shape}")
print(f"Dimensiones: [C={audio_tensor.shape[0]}, L={audio_tensor.shape[1]}]")
print()

# ============================================================================
# FASE B: INTERVENCIÓN QUIRÚRGICA
# ============================================================================
print("="*80)
print("FASE B: INTERVENCIÓN QUIRÚRGICA")
print("="*80)

# hace un clon para no arruinar la imagen original y poder comparar después
img_tensor_ablated = img_tensor.clone()

#se elimina toda la información del canal verde: G(x,y) = 0 ∀(x,y)
img_tensor_ablated[1, :, :] = 0
# flip() invierte el eje especificado: dims=[1] es el eje temporal (L).
audio_tensor_reversed = torch.flip(audio_tensor, dims=[1])

print(f"✅ Ablación completada. Canal verde suma: {img_tensor_ablated[1].sum():.6f}")
print(f"✅ Audio invertido temporalmente")
print()

# ============================================================================
# FASE C: RECONSTRUCCIÓN PERCEPTUAL
# ============================================================================
print("="*80)
print("FASE C: RECONSTRUCCIÓN PERCEPTUAL")
print("="*80)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# permute() reordena dimensiones: [C,H,W] → [H,W,C]
# Matplotlib necesita formato [altura, ancho, canales] para imshow()
img_display_original = img_tensor.permute(1, 2, 0).cpu().numpy()
# clip() limita valores al rango [0,1] para prevenir errores de visualización
img_display_original = np.clip(img_display_original, 0, 1)
axes[0].imshow(img_display_original)
axes[0].set_title('ORIGINAL (RGB)', fontweight='bold', fontsize=14)
axes[0].axis('off')

img_display_ablated = img_tensor_ablated.permute(1, 2, 0).cpu().numpy()
img_display_ablated = np.clip(img_display_ablated, 0, 1)
axes[1].imshow(img_display_ablated)
axes[1].set_title('POST-ABLACIÓN (G=0)', fontweight='bold', fontsize=14, color='red')
axes[1].axis('off')

plt.suptitle('DISECCIÓN DIGITAL: Ablación del Canal Verde', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

print("\n🎧 Audio Original (440 Hz):")
display(Audio(audio_tensor.cpu().numpy(), rate=sample_rate))

print("\n🎧 Audio Invertido Temporalmente:")
display(Audio(audio_tensor_reversed.cpu().numpy(), rate=sample_rate))
```

# ***De Señal Física a Tensor Matemático***

 **Imagen RGB**
- **Entrada**: Imagen de 256×256 píxeles con 3 canales de color
- **Proceso**: Conversión a tensor de punto flotante normalizado
[0.0, 1.0]
- **Formato PyTorch**:`[C, H, W]` =` [Canales, Alto, Ancho]`


**Dimensiones**:` [C=3, H=256, W=256]`



---


**Interpretacion fisica de los ejes:**



*   C = canales, son tres componentes espectrales del modelo (RGB)

*   Canal 0 (ROJO) : Intensidad o presencia de luz en longitudes de onda entre 620 a 750 nm.

*   Canal 1 (VERDE): Intensidad o presencia de luz en longitudes de onda entre 495 a 570 nm.

*   Canal 2 (AZUL): Intensidad o presencia de luz en longitudes de onda entre 450 a 495 nm.

*   H (HEIGHT): 256, coordenada vertical (Pixeles en el eje Y)


*   W (WIDTH): 256, coordenada horizontal (Pixeles en el eje X)

Cada elemento `tensor[c, h, w]` codifica la intensidad del color `c` captada por el sensor en la posición espacial `(h, w)`.


---



**Audio Estéreo**
- **Entrada:** Señal de audio sintética (tono puro de 440 Hz - La4)
- **Parámetros:** 16,000 muestras/segundo, 2 segundos de duración
- **Formato PyTorch:** `[C, L]` = `[Canales, Longitud]`

**Interpretación Física de los Ejes:**
- **C (Channels) = 2:** Canales espaciales de audio estereofónico
  - Canal 0: Transductor IZQUIERDO (oído/altavoz izquierdo)
  - Canal 1: Transductor DERECHO (oído/altavoz derecho)
  
- **L (Length) = 32,000:** Longitud temporal discreta
  - Número de muestras totales
  - Resolución temporal: 0.0625 ms por muestra (1/16000 segundos)
  - Duración total: 2.0 segundos

Cada elemento `tensor[c, l]` codifica la amplitud de presión sonora en el canal `c` en el instante temporal discreto `l`.

**análisis fenomenológico**

**¿Por qué al multiplicar la matriz por un escalar mayor a 1 la imagen se percibe más "brillante"?**

**Nivel Matemático**
Multiplicar por α > 1 aumenta el brillo porque:
`nuevo_valor = α × valor_original`
Cada pixel aumenta proporcionalmente

**Nivel fisico**
Los valores del tensor representan intensidad lumínica:

0.0 = Sin luz (negro)
1.0 = Máxima luz (blanco)
Multiplicar × 1.5 = Simular 50% más fotones llegando al sensor

**Nivel percepcion**
Es decirle a la pantalla que emita mas energia a cada pixel, lo que el ojo interpreta como una imagen mas clara o mas brillante


**¿Por qué al eliminar el verde la imagen se ve Magenta?**

**Nivel matematico**
[R, G, B] → [R, 0, B]
Solo quedan dos canales activos: Rojo y Azul.

**Nivel percepcion**
Si apagas el canal verde (G=0), solo quedan el rojo y el azul. Al mezclarse estos dos colores en nuestro cerebro sin la presencia del verde, percibimos el color Magenta. Al eliminar el verde o mejor dicho en ausencia del verde y presencia del rojo y azul magenta absorbe su color complementario verde haciendolo ausente de luz.
